```python
# Traçar os seguintes sinais discretos no tempo: xa, xb e xc
# Antes de traçar os sinais, definir as funções correspondentes a cada sinal e, 
# em seguida, utilizar uma biblioteca de plotagem, como Matplotlib, para visualizá-los no tempo.

import numpy as np
import matplotlib.pyplot as plt 

def x(n, a):
    # Definindo o sinal xa[n] = a^n 
    return float(a) ** n


# base de tempo discreto
n = np.arange(-10, 11, 1)  # Intervalo de n

a = -0.5  # Valor de a para o sinal xa[n]   

plt.figure(figsize=(12, 8))
# Traçando o sinal xa[n]    
plt.subplot(3, 1, 1)
plt.stem(n, x(n, a))
plt.title('Sinal xa[n] = a^n')
plt.xlabel('n')
plt.ylabel('xa[n]')
plt.grid(True)  

b = 2**(-1)  # Valor de b para o sinal xb[n]   
 
# Traçando o sinal xb[n]    
plt.subplot(3, 1, 2)
plt.stem(n, x(n, b))
plt.title('Sinal xb[n] = b^n')
plt.xlabel('n')
plt.ylabel('xb[n]')
plt.grid(True)  

c = -2  # Valor de c para o sinal xc[n]   
 
# Traçando o sinal xc[n]    
plt.subplot(3, 1, 3)
plt.stem(n, x(n, c))
plt.title('Sinal xc[n] = c^n')
plt.xlabel('n')
plt.ylabel('xc[n]')
plt.grid(True)  
plt.xlabel('n')
plt.ylabel('xb[n]')
plt.grid(True)  
```

```python
# x = cos(pi/12*n + pi/4)  # Valor de x para o sinal xd[n]
from numpy import pi, cos
import matplotlib.pyplot as plt


n = range(-30,30)  # Intervalo Inteiro de n
w=pi/12  # Frequência angular
ang = [w*ni + pi/4 for ni in n]
x = [cos(a) for a in ang]
plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.stem(n, x)
plt.title('Sinal xd[n] = cos(w*n + pi/4)')
plt.xlabel('n')
plt.ylabel('xd[n]')

w =2
ang = [w*ni + pi/4 for ni in n]
x = [cos(a) for a in ang]
plt.subplot(3, 1, 2)
plt.stem(n, x)
plt.title('Sinal xd[n] = cos(w*n + pi/4)')
plt.xlabel('n')
plt.ylabel('xd[n]')         
```
