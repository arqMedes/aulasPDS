```python
# Exercício 3M-8
import numpy as np

def media_movel_numpy(vetor, n):
    # O modo 'valid' garante que o cálculo só ocorra onde há N pontos cheios
    return np.convolve(vetor, np.ones(n), mode='valid') / n
def parametros_media_movel(N):
    """
    Retorna os vetores b e a para implementar um filtro de 
    média móvel de N pontos através de funções de filtragem.
    """
    # b é um vetor preenchido com 1/N repetido N vezes
    b = np.ones(N) / N
    
    # a é apenas [1], já que é um filtro FIR (sem feedback da saída)
    a = np.array([1.0])
    
    return b, a

# --- Exemplo prático de uso ---
N_pontos = 4
b, a = parametros_media_movel(N_pontos)

print("Vetor b:", b)
print("Vetor a:", a)
# Exemplo de uso:
dados = [10, 20, 30, 40, 50, 60]
print(media_movel_numpy(dados, 3))
# Saída: [20. 30. 40. 50.]

```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definição do vetor de tempo discreto (tamanho 45)
n = np.arange(45)

# 2. Construção do sinal de entrada: x[n] = cos(πn/5) + δ[n − 30] − δ[n − 35]
x = np.cos(np.pi * n / 5)
x[30] += 1.0  # Impulso unitário positivo em n=30
x[35] -= 1.0  # Impulso unitário negativo em n=35

def filtrar_media_movel(sinal, N):
    """Aplica o filtro de média móvel usando convolução do NumPy."""
    b = np.ones(N) / N
    # A convolução com 'full' estende o sinal; truncamos em [:len(sinal)] 
    # para simular o comportamento de regime transiente inicial zerado
    return np.convolve(sinal, b, mode='full')[:len(sinal)]

# 3. Processamento para os diferentes valores de N
y_4 = filtrar_media_movel(x, 4)
y_8 = filtrar_media_movel(x, 8)
y_12 = filtrar_media_movel(x, 12)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definição do vetor de tempo discreto (tamanho 45)
n = np.arange(45)

# 2. Construção do sinal de entrada
x = np.cos(np.pi * n / 5)
x[30] += 1.0  # Kronecker delta em n=30
x[35] -= 1.0  # Kronecker delta em n=35

def filtrar_media_movel(sinal, N):
    b = np.ones(N) / N
    # Retorna apenas o tamanho original simulando condições iniciais nulas
    return np.convolve(sinal, b, mode='full')[:len(sinal)]

# 3. Gerar e salvar um gráfico separado para cada N
for N in [4, 8, 12]:
    y = filtrar_media_movel(x, N)
    
    plt.figure(figsize=(10, 4))
    # Sinal original em azul
    plt.stem(n, x, linefmt='b-', markerfmt='bo', basefmt='k-', label='Original x[n]')
    # Sinal filtrado em vermelho com linha tracejada
    plt.stem(n, y, linefmt='r--', markerfmt='ro', basefmt='k-', label=f'Filtrado y[n] (N={N})')
    
    plt.title(f'Filtro de Média Móvel de {N} Pontos')
    plt.xlabel('Amostras (n)')
    plt.ylabel('Amplitude')
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend()
    plt.tight_layout()
    
    # Salva a imagem localmente
    plt.savefig(f'filtro_media_movel_N{N}.png', dpi=150)
    plt.close()
```
